
# FieldShift — End-to-End Pipeline Execution

## Objective

Execute and inspect the FieldShift pipeline.

Pipeline stages:

P1 — Grid Layer Preparation

P2 — Crop Simulation

P3 — Monitoring

P4 — Learning

P5 — Crop Optimization


```python

from pathlib import Path
import sys
import json
from pprint import pprint

ROOT = Path.cwd()

if not (ROOT / "pipelines").exists():
    ROOT = ROOT.parent

sys.path.insert(0, str(ROOT))

from pipelines.compile import (
    run_fieldshift_pipeline,
    FieldShiftPipelineError
)

print("FieldShift pipeline imported successfully.")
print("Project root:", ROOT)
```

```python

grid_input = None

crops = []

simulator = None

simulation_config = {}

monitor_checks = {}

monitor_inputs = {}

baseline_scores = {}

farmer_feedback = {}

candidate_data = []

optimization_constraints = None

historical_scores = None

historical_outcomes = None

objective = "score"
```

```python

required = {
    "grid_input": grid_input,
    "crops": crops,
    "simulator": simulator,
    "candidate_data": candidate_data,
    "optimization_constraints": optimization_constraints
}

missing = [
    name
    for name, value in required.items()
    if value is None or value == []
]

if missing:

    print("Missing required inputs:", missing)

    print(
        "Configure actual simulator, grid, candidates, "
        "and optimization constraints before execution."
    )

else:

    try:

        result = run_fieldshift_pipeline(
            grid_input=grid_input,
            crops=crops,
            simulator=simulator,
            simulation_config=simulation_config,
            monitor_checks=monitor_checks,
            monitor_inputs=monitor_inputs,
            baseline_scores=baseline_scores,
            farmer_feedback=farmer_feedback,
            candidate_data=candidate_data,
            optimization_constraints=optimization_constraints,
            historical_scores=historical_scores,
            historical_outcomes=historical_outcomes,
            objective=objective
        )

        print("Pipeline Status:", result["status"])

        pprint(result)

    except FieldShiftPipelineError as error:

        print("Pipeline Execution Failed:", error)

        raise
```

```python

if (
    "result" in globals()
    and isinstance(result, dict)
    and result.get("status") == "success"
):

    OUTPUT_DIR = ROOT / "outputs"

    OUTPUT_DIR.mkdir(exist_ok=True)

    OUTPUT_FILE = (
        OUTPUT_DIR / "fieldshift_notebook_result.json"
    )

    OUTPUT_FILE.write_text(
        json.dumps(result, indent=2, default=str),
        encoding="utf-8"
    )

    print("Output saved:", OUTPUT_FILE)

else:

    print("No successful pipeline result available.")
```
