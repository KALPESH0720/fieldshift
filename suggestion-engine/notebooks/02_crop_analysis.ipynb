
# FieldShift — Feature Engineering Checks

## Objective

Inspect the FieldShift feature engineering modules and validate feature inputs and outputs.

This notebook focuses on:
- Feature module inspection
- Pilot dataset loading
- Feature implementation review
- Feature validation


```python

from pathlib import Path
import sys
import json
import pandas as pd

ROOT = Path.cwd()

if not (ROOT / "src").exists():
    ROOT = ROOT.parent

sys.path.insert(0, str(ROOT / "src"))

FEATURE_DIR = ROOT / "src" / "fieldshift" / "features"

print("Project root:", ROOT)
print("Feature directory:", FEATURE_DIR)
```

```python

feature_files = sorted(
    file.name
    for file in FEATURE_DIR.glob("*.py")
)

print("Available Feature Modules:")

for file in feature_files:
    print(file)
```

```python

with open(
    ROOT / "tests" / "fixtures" / "pilot_cells.json",
    encoding="utf-8"
) as f:

    raw = json.load(f)

if isinstance(raw, list):
    records = raw

elif isinstance(raw, dict):
    records = raw.get("cells", [])

else:
    records = []

df = pd.DataFrame(records)

display(df.head())

print("Dataset Shape:", df.shape)
```

```python

for file in sorted(FEATURE_DIR.glob("*.py")):

    if file.name == "__init__.py":
        continue

    print("\n---", file.name, "---")

    print(
        file.read_text(encoding="utf-8")
    )
```

```python

print("Checking dataset...")

print("\nMissing Values:")
display(df.isnull().sum())

print("\nNumeric Columns:")
display(df.select_dtypes(include="number").describe())

print("\nFeature Engineering Validation Completed.")
```


## Conclusion

Existing FieldShift feature modules were inspected.

The pilot dataset was loaded and checked for missing values and numeric characteristics.

Feature implementations can now be connected to the pipeline using their actual function definitions.
