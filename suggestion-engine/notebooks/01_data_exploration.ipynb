
# FieldShift — Pilot Data Exploration

## Objective
Explore pilot-region and pilot-cell datasets before running the FieldShift pipelines.

This notebook examines:
- Pilot region information
- Pilot grid cells
- Missing values
- Geographic coordinates
- Basic dataset visualization


```python

from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd()

if not (ROOT / "tests").exists():
    ROOT = ROOT.parent

REGION_FILE = ROOT / "tests" / "fixtures" / "pilot_region.json"
CELLS_FILE = ROOT / "tests" / "fixtures" / "pilot_cells.json"

print("Project root:", ROOT)
print("Region fixture exists:", REGION_FILE.exists())
print("Cells fixture exists:", CELLS_FILE.exists())
```

```python

with REGION_FILE.open(encoding="utf-8") as f:
    region = json.load(f)

with CELLS_FILE.open(encoding="utf-8") as f:
    cells_raw = json.load(f)

print("Pilot Region:")
print(region)

print("\nPilot Cells:")
print(cells_raw)
```

```python

if isinstance(cells_raw, list):
    records = cells_raw

elif isinstance(cells_raw, dict):
    records = cells_raw.get("cells", [])

else:
    records = []

cells = pd.DataFrame(records)

display(cells.head())

print("Number of cells:", len(cells))
print("Columns:", list(cells.columns))
```

```python

print("Missing values:")
display(cells.isnull().sum())

print("\nDataset Information:")
display(cells.describe(include="all"))
```

```python

if {"latitude", "longitude"}.issubset(cells.columns):

    plt.figure(figsize=(8, 6))

    plt.scatter(
        cells["longitude"],
        cells["latitude"],
        alpha=0.75
    )

    plt.xlabel("Longitude")
    plt.ylabel("Latitude")

    plt.title("FieldShift Pilot Grid Cells")

    plt.grid(True)
    plt.show()

else:
    print("Latitude and longitude columns not found.")
```


## Conclusion

Pilot-region and pilot-cell datasets were loaded successfully.

Dataset structure, missing values, and geographic distribution were inspected.

The dataset is now ready for further validation and pipeline processing.
